```
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO
from google.colab import files
from PIL import Image

# ==========================================
# PHẦN 1: CÁC HÀM XỬ LÝ ẢNH (CẮT & XOAY PCB)
# ==========================================

def order_points(pts):
    # Sắp xếp 4 điểm tọa độ: Trái-Trên, Phải-Trên, Phải-Dưới, Trái-Dưới
    rect = np.zeros((4, 2), dtype="float32")
    s = pts.sum(axis=1)
    rect[0] = pts[np.argmin(s)]
    rect[2] = pts[np.argmax(s)]
    diff = np.diff(pts, axis=1)
    rect[1] = pts[np.argmin(diff)]
    rect[3] = pts[np.argmax(diff)]
    return rect

def four_point_transform(image, pts):
    # Biến đổi phối cảnh để làm thẳng ảnh
    rect = order_points(pts)
    (tl, tr, br, bl) = rect

    widthA = np.sqrt(((br[0] - bl[0]) ** 2) + ((br[1] - bl[1]) ** 2))
    widthB = np.sqrt(((tr[0] - tl[0]) ** 2) + ((tr[1] - tl[1]) ** 2))
    maxWidth = max(int(widthA), int(widthB))

    heightA = np.sqrt(((tr[0] - br[0]) ** 2) + ((tr[1] - br[1]) ** 2))
    heightB = np.sqrt(((tl[0] - bl[0]) ** 2) + ((tl[1] - bl[1]) ** 2))
    maxHeight = max(int(heightA), int(heightB))

    dst = np.array([
        [0, 0],
        [maxWidth - 1, 0],
        [maxWidth - 1, maxHeight - 1],
        [0, maxHeight - 1]], dtype="float32")

    M = cv2.getPerspectiveTransform(rect, dst)
    warped = cv2.warpPerspective(image, M, (maxWidth, maxHeight))
    return warped

def crop_pcb(image):
    # 1. Chuyển sang ảnh xám và làm mờ
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # 2. Phát hiện biên Canny
    v = np.median(blurred)
    lower = int(max(0, (1.0 - 0.33) * v))
    upper = int(min(255, (1.0 + 0.33) * v))
    edged = cv2.Canny(blurred, lower, upper)

    # 3. Tìm Contour lớn nhất
    contours, _ = cv2.findContours(edged.copy(), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if len(contours) > 0:
        c = max(contours, key=cv2.contourArea)

        # Nếu diện tích contour quá nhỏ (nhiễu), bỏ qua
        if cv2.contourArea(c) < 1000:
            return None

        # 4. Tìm hình chữ nhật bao quanh
        rect = cv2.minAreaRect(c)
        box = cv2.boxPoints(rect)

        # --- SỬA LỖI TẠI ĐÂY ---
        # Thay np.int0 bằng np.int32 hoặc ép kiểu int
        box = np.int32(box)
        # -----------------------

        # 5. Cắt và xoay
        warped = four_point_transform(image, box.reshape(4, 2))

        # 6. Chuẩn hóa chiều (Luôn nằm ngang)
        h, w = warped.shape[:2]
        if h > w:
            warped = cv2.rotate(warped, cv2.ROTATE_90_CLOCKWISE)

        return warped
    return None

# ==========================================
# PHẦN 2: MAIN PROGRAM (DETECT & HIỂN THỊ)
# ==========================================

# 1. Load Model (Đảm bảo file best.pt tồn tại)
model_path = 'runs/detect/train/weights/best.pt' # Kiểm tra lại đường dẫn này trên Colab của bạn
try:
    model = YOLO(model_path)
    print("✅ Đã load model thành công!")
except:
    print("❌ Không tìm thấy model! Vui lòng kiểm tra lại đường dẫn file best.pt")

# 2. Upload ảnh
print("\n📸 Vui lòng chọn ảnh PCB để test:")
uploaded = files.upload()

for filename in uploaded.keys():
    # Đọc ảnh gốc
    original_img = cv2.imread(filename)

    # Xử lý cắt PCB
    print(f"ang xử lý ảnh: {filename}...")
    cropped_pcb = crop_pcb(original_img)

    if cropped_pcb is not None:
        # --- TRƯỜNG HỢP 1: CẮT ĐƯỢC PCB ---
        print("✅ Đã tách được PCB khỏi nền!")

        # Chạy model trên ảnh ĐÃ CẮT (Kết quả sẽ chính xác hơn)
        results = model.predict(cropped_pcb, conf=0.25)

        # Vẽ box lên ảnh đã cắt (dùng plot() của YOLO để lấy array ảnh kết quả)
        result_plot = results[0].plot()

        # Hiển thị 3 hình: Gốc -> Đã cắt -> Kết quả Detect
        fig, axes = plt.subplots(1, 3, figsize=(18, 6))

        # Ảnh gốc
        axes[0].imshow(cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB))
        axes[0].set_title("1. Ảnh gốc (Có nền)")
        axes[0].axis('off')

        # Ảnh sau khi crop & align
        axes[1].imshow(cv2.cvtColor(cropped_pcb, cv2.COLOR_BGR2RGB))
        axes[1].set_title("2. Đã cắt & Xoay thẳng")
        axes[1].axis('off')

        # Kết quả Detect
        axes[2].imshow(cv2.cvtColor(result_plot, cv2.COLOR_BGR2RGB))
        axes[2].set_title("3. Kết quả YOLO Detect")
        axes[2].axis('off')

        plt.show()

    else:
        # --- TRƯỜNG HỢP 2: KHÔNG TÌM THẤY BIÊN (DETECT TRÊN ẢNH GỐC) ---
        print("⚠️ Không tách được nền (do độ tương phản thấp?). Detect trực tiếp trên ảnh gốc.")

        results = model.predict(original_img, conf=0.25)
        result_plot = results[0].plot()

        plt.figure(figsize=(10, 6))
        plt.imshow(cv2.cvtColor(result_plot, cv2.COLOR_BGR2RGB))
        plt.title("Kết quả Detect (Ảnh gốc)")
        plt.axis('off')
        plt.show()
```
